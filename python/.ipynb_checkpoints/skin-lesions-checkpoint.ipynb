```python
# Funkcja do unzipowania zdjęć
def unzip(zip_path, extract_to):
    if not os.path.exists(zip_path):
        print(f"ZIP nie istnieje: {zip_path}")
        return
    
    os.makedirs(extract_to, exist_ok=True)
    
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_to)
```

```python
# Usunięcie uszkodzonych obrazów ze zbiorów
def remove_corrupted(metadata, path_column="path"):
    good_paths = []
    bad_paths = []

    for path in metadata[path_column]:
        try:
            if path is None or not os.path.exists(path):
                bad_paths.append(path)
                continue

            img = Image.open(path)
            img.verify()
            good_paths.append(path)

        except Exception:
            bad_paths.append(path)

    return metadata[metadata[path_column].isin(good_paths)].reset_index(drop=True)
```

```python
# Wyświetlenie wszystkich podklas w każdej klasie
reverse_labels = defaultdict(list)
for orig, new in labels.items():
    reverse_labels[new].append(orig)

COLS = 4

for new_class, original_list in reverse_labels.items():
    subset = metadata_all[metadata_all["dx"] == new_class]
    
    examples = []
    for orig in original_list:
        row = subset[subset["original_dx"] == orig].sample(1)
        if len(row) > 0:
            examples.append(row.iloc[0])

    n = len(examples)
    rows = math.ceil(n / COLS)

    fig, axes = plt.subplots(rows, COLS, figsize=(COLS * 4, rows * 4))
    axes = axes.flatten()

    if n < COLS:
        offset = (COLS - n) // 2
    else:
        offset = 0

    for i, ax in enumerate(axes):
        if i < offset or i >= offset + n:
            ax.axis("off")
            continue

        row = examples[i - offset]
        img = Image.open(row["path"])

        ax.imshow(img)
        ax.axis("off")
        ax.set_title(row["original_dx"], fontsize=10)

    plt.suptitle(f"{new_class}", fontsize=16, fontweight="bold")
    plt.tight_layout()
    plt.show()
```

```python
# Oversampling klas rzadkich w zbiorze dermoskopii
target_size = 5000

balanced_frames = []

for label in metadata_derm['dx'].unique():
    df_class = metadata_derm[metadata_derm["dx"] == label]
    count = len(df_class)

    if count < target_size:
        df_upsampled = resample(
            df_class,
            replace=True,
            n_samples=target_size,
            random_state=42
        )
        balanced_frames.append(df_upsampled)
    else:
        df_downsampled = df_class.sample(n=target_size, random_state=42)
        balanced_frames.append(df_downsampled)

metadata_derm = pd.concat(balanced_frames).sample(frac=1, random_state=42).reset_index(drop=True)
print(metadata_derm["dx"].value_counts())
```

```python
# Wizualizacja po oversamplingu
class_counts = metadata_derm['dx'].value_counts()

labels = class_counts.index
sizes = class_counts.values

plt.figure(figsize=(12, 10))
plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    startangle=140,
    textprops={'fontsize': 10}
)

plt.title("Procentowy udział klas po oversamplingu", fontsize=20, pad=30)
plt.axis('equal')
plt.tight_layout()
plt.show()
```

# Główna klasa do trenowania modelu

```python
class DomainShiftCallback(tf.keras.callbacks.Callback):
    """
    Callback monitorujący różnicę jakości między domeną telefonu i dermoskopii po każdej epoce (na zbiorach walidacyjnych).
    """
    def __init__(self, val_phone_gen, val_derm_gen):
        super().__init__()
        self.val_phone_gen = val_phone_gen
        self.val_derm_gen = val_derm_gen

    def on_epoch_end(self, epoch, logs=None):
        phone_eval = self.model.evaluate(self.val_phone_gen, verbose=0)
        derm_eval = self.model.evaluate(self.val_derm_gen, verbose=0)

        phone_acc = phone_eval[1]
        derm_acc = derm_eval[1]

        diff = derm_acc - phone_acc

        print(f"\n Phone ACC: {phone_acc:.4f} | Derm ACC: {derm_acc:.4f} | delta = {diff:.4f}")

        if abs(diff) > 0.10:
            print("Domain shift rośnie")
```

```python
class TransferLearningTuner:
    def __init__(
        self,
        model_name,
        num_classes,
        train_gen,
        val_gen,
        test_gen,
        val_phone_gen=None,
        val_derm_gen=None,
        hidden_units=[128],
        lr_frozen=1e-4,
        lr_unfrozen=2e-5,
        epochs_frozen=3,
        epochs_unfrozen=7,
        dropout_rate=0.3
    ):
        self.model_name = model_name
        self.num_classes = num_classes
        self.train_gen = train_gen
        self.val_gen = val_gen
        self.test_gen = test_gen
        self.val_phone_gen = val_phone_gen
        self.val_derm_gen = val_derm_gen
        self.hidden_units = hidden_units
        self.lr_frozen = lr_frozen
        self.lr_unfrozen = lr_unfrozen
        self.epochs_frozen = epochs_frozen
        self.epochs_unfrozen = epochs_unfrozen
        self.dropout_rate = dropout_rate

        self.model = self.build_model()

    def build_model(self):
        base = MODELS[self.model_name](
            weights="imagenet",
            include_top=False,
            input_shape=(224, 224, 3)
        )

        base.trainable = False
        self.base = base

        x = tf.keras.layers.GlobalAveragePooling2D()(base.output)

        for units in self.hidden_units:
            x = tf.keras.layers.Dense(units, activation="relu")(x)
            x = tf.keras.layers.BatchNormalization()(x)
            x = tf.keras.layers.Dropout(self.dropout_rate)(x)

        output = tf.keras.layers.Dense(self.num_classes, activation="softmax")(x)

        return tf.keras.Model(inputs=base.input, outputs=output)

    def train_frozen(self):
        self.model.compile(
            optimizer=tf.keras.optimizers.Adam(self.lr_frozen),
            loss=tf.keras.losses.CategoricalFocalCrossentropy(gamma=2.0, alpha=0.25),
            metrics=["accuracy", tf.keras.metrics.AUC()]
        )

        ckpt = tf.keras.callbacks.ModelCheckpoint(
            f"{self.model_name}_frozen.weights.h5",
            monitor="val_loss",
            save_best_only=True,
            save_weights_only=True
        )

        early_stopping = tf.keras.callbacks.EarlyStopping(
            monitor="val_loss", patience=5, restore_best_weights=True
        )

        reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.5,
            patience=2,
            cooldown=2,
            min_lr=1e-7,
            verbose=1
        )

        history = self.model.fit(
            self.train_gen,
            validation_data=self.val_gen,
            epochs=self.epochs_frozen,
            callbacks=[ckpt, early_stopping, reduce_lr]
        )

        self.model.load_weights(f"{self.model_name}_frozen.weights.h5")
        return history

    def train_unfrozen(self):
        fine_tune_at = int(len(self.base.layers) * 0.7)

        for i, layer in enumerate(self.base.layers):
            if i < fine_tune_at or isinstance(layer, tf.keras.layers.BatchNormalization):
                layer.trainable = False
            else:
                layer.trainable = True

        self.model.compile(
            optimizer=tf.keras.optimizers.Adam(self.lr_unfrozen),
            loss=tf.keras.losses.CategoricalFocalCrossentropy(gamma=2.0, alpha=0.25),
            metrics=["accuracy", tf.keras.metrics.AUC()]
        )

        ckpt = tf.keras.callbacks.ModelCheckpoint(
            f"{self.model_name}_unfrozen.weights.h5",
            monitor="val_loss",
            save_best_only=True,
            save_weights_only=True
        )

        early_stopping = tf.keras.callbacks.EarlyStopping(
            monitor="val_loss", patience=5, restore_best_weights=True
        )

        reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.5,
            patience=2,
            cooldown=2,
            min_lr=1e-7,
            verbose=1
        )

        history = self.model.fit(
            self.train_gen,
            validation_data=self.val_gen,
            epochs=self.epochs_unfrozen,
            callbacks=[ckpt, early_stopping, reduce_lr]
        )

        self.model.load_weights(f"{self.model_name}_unfrozen.weights.h5")
        return history

    def evaluate(self):
        return self.model.evaluate(self.test_gen)

    def plot_confusion_matrix(self, y_true, y_pred, labels, title):
        cm = confusion_matrix(y_true, y_pred)
        cm_norm = cm.astype("float") / cm.sum(axis=1)[:, np.newaxis]

        plt.figure(figsize=(10, 8))
        sns.heatmap(cm_norm, annot=True, fmt=".2f", cmap="Blues",
                    xticklabels=labels, yticklabels=labels)
        plt.title(title)
        plt.ylabel("True class")
        plt.xlabel("Predicted class")
        plt.show()

    def predict_generator(self, generator):
        preds = self.model.predict(generator, steps=len(generator), verbose=0)
        y_true = generator.classes
        y_pred = np.argmax(preds, axis=1)
        return y_true, y_pred

    def compute_f1_per_class(self, y_true, y_pred, labels):
        y_true = np.array(y_true)
        y_pred = np.array(y_pred)

        f1 = f1_score(
            y_true,
            y_pred,
            average=None,
            labels=list(range(len(labels))),
            zero_division=0
        )

        df = pd.DataFrame({"class": labels, "f1": f1})
        return df

    def plot_loss(self, history, name):
        """
        Drukuje wykres straty i dokładności dla treningu i walidacji
        """
        plt.figure(figsize=(12, 8))
        epochs = range(1, len(history.history["loss"]) + 1)
        
        plt.plot(epochs, history.history["loss"], label="train_loss")
        plt.plot(epochs, history.history["val_loss"], label="val_loss")
        plt.plot(epochs, history.history["accuracy"], label="train_acc")
        plt.plot(epochs, history.history["val_accuracy"], label="val_acc")
        
        plt.title(f"Training Loss and Accuracy – {name}")
        plt.xlabel("Epoka")
        plt.ylabel("Loss / Accuracy")
        plt.legend()
        plt.show()
        
    def run(self):
        print(f"\n{self.model_name}: Etap 1 (Frozen)")
        history_frozen = self.train_frozen()

        print(f"\n{self.model_name}: Etap 2 (Fine-tuning)")
        history_unfrozen = self.train_unfrozen()

        print("\n=== Wyniki ===")
        results = self.evaluate()

        labels = list(self.train_gen.class_indices.keys())

        y_true_phone, y_pred_phone = self.predict_generator(self.val_phone_gen)
        f1_phone = self.compute_f1_per_class(y_true_phone, y_pred_phone, labels)

        y_true_derm, y_pred_derm = self.predict_generator(self.val_derm_gen)
        f1_derm = self.compute_f1_per_class(y_true_derm, y_pred_derm, labels)

        mean_f1_phone = float(f1_phone["f1"].mean())
        mean_f1_derm = float(f1_derm["f1"].mean())
        domain_shift = mean_f1_derm - mean_f1_phone

        return {
            "results": results,
            "history_frozen": history_frozen,
            "history_unfrozen": history_unfrozen,
            "f1_phone": f1_phone,
            "f1_derm": f1_derm,
            "mean_f1_phone": mean_f1_phone,
            "mean_f1_derm": mean_f1_derm,
            "domain_shift": domain_shift
        }
```

```python
all_results = []
```

# ResNet50

```python
MODEL_NAME = "ResNet50"

tuner = TransferLearningTuner(
    model_name=MODEL_NAME,
    num_classes=NUM_CLASSES,
    train_gen=train_generator,
    val_gen=val_derm_generator,
    test_gen=test_generator,
    val_phone_gen=val_phone_generator,
    val_derm_gen=val_derm_generator,
    hidden_units=[512, 128],
    epochs_frozen=5,
    epochs_unfrozen=5
)

out = tuner.run()

results = out["results"]
history_frozen = out["history_frozen"]
history_unfrozen = out["history_unfrozen"]
f1_phone = out["f1_phone"]
f1_derm = out["f1_derm"]
mean_f1_phone = out["mean_f1_phone"]
mean_f1_derm = out["mean_f1_derm"]
domain_shift = out["domain_shift"]
```

```python
tuner.plot_loss(history_frozen, "Frozen")
tuner.plot_loss(history_unfrozen, "Unfrozen")
```

```python
labels = list(train_generator.class_indices.keys())

f1_phone = out["f1_phone"]
f1_derm  = out["f1_derm"]

f1_phone["f1"] = f1_phone["f1"].replace(0, "N/A")
df = pd.DataFrame({
    "class": labels,
    "F1_phone": f1_phone["f1"],
    "F1_derm":  f1_derm["f1"]
})

print(tabulate(df, headers="keys", tablefmt="fancy_grid", floatfmt=".4f"))
```

```python
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

tuner.model.export(f"models/skin_lesion_model_{MODEL_NAME}")

converter = tf.lite.TFLiteConverter.from_saved_model(
    f"models/skin_lesion_model_{MODEL_NAME}"
)
converter.optimizations = [tf.lite.Optimize.DEFAULT]

tflite_model = converter.convert()
with open(f"tflite_models/skin_lesion_model_{MODEL_NAME}.tflite", "wb") as f:
    f.write(tflite_model)
```

```python
f1_phone_numeric = pd.to_numeric(f1_phone["f1"], errors="coerce")
f1_derm_numeric  = pd.to_numeric(f1_derm["f1"], errors="coerce")

mean_f1_phone = f1_phone_numeric.mean()
mean_f1_derm  = f1_derm_numeric.mean()

domain_shift = mean_f1_derm - mean_f1_phone

all_results.append({
    "model": MODEL_NAME,
    "test_loss": results[0],
    "test_acc": results[1],
    "test_auc": results[2],
    "f1_phone": f1_phone,
    "f1_derm": f1_derm,
    "domain_shift": domain_shift
})
```
